```python
import numpy as np
import matplotlib.pyplot as plt

# ----- Data from your table -----
methods = [
    "Poisoned RAG",
    "Paraphrasing",
    "Knowledge\nExpansion (k=10)",
    "Knowledge\nExpansion (k=25)",
    "Duplicate-\nFiltering",
    "KG-Validated\nRAG",
]

asr = [97.0, 89.0, 89.0, 69.0, 97.0, 30.0]
correct = [1.0, 9.0, 11.0, 17.0, 1.0, 56.0]
other_incorrect = [0.0, 1.0, 0.0, 14.0, 0.0, 14.0]
answer_unavail = [2.0, 1.0, 0.0, 0.0, 2.0, 0.0]

# ----- Plot settings -----
x = np.arange(len(methods))        # positions for each method
width = 0.2                        # width of each bar

fig, ax = plt.subplots(figsize=(10, 5))

# four bar groups
bars_asr = ax.bar(x - 1.5*width, asr, width, label="ASR")
bars_correct = ax.bar(x - 0.5*width, correct, width, label="Correctness")
bars_other = ax.bar(x + 0.5*width, other_incorrect, width, label="Other Incorrect")
bars_unavail = ax.bar(x + 1.5*width, answer_unavail, width, label="Answer Unavailability")

# labels and layout
ax.set_ylabel("Percentage (%)")
ax.set_xticks(x)
ax.set_xticklabels(methods, rotation=20, ha="right")
ax.set_ylim(0, 110)  # a bit above 100 for headroom
ax.legend()
ax.set_title("Main Results by Method")

plt.tight_layout()

# save to file for LaTeX
plt.savefig("main_results_bar_chart.png", dpi=300, bbox_inches="tight")

# or show interactively
plt.show()
```

```python
import matplotlib.pyplot as plt

# ----- Data from your table -----
sizes = ["5k", "10k", "15k", "20k"]   # 4 labels
x = [5, 10, 15, 20]                   # 4 numeric positions

asr = [57.0, 42.0, 30.0, 28.0]
correct = [31.0, 47.0, 56.0, 58.0]
other_incorrect = [11.0, 10.0, 14.0, 14.0]
answer_unavail = [1.0, 1.0, 0.0, 0.0]

plt.figure(figsize=(8, 4))

plt.plot(x, asr, marker="o", label="ASR (%)")
plt.plot(x, correct, marker="o", label="Correctness (%)")
plt.plot(x, other_incorrect, marker="o", label="Other Incorrect (%)")
plt.plot(x, answer_unavail, marker="o", label="Answer Unavailability (%)")

plt.xticks(x, sizes)
plt.xlabel("Dataset Size (documents)")
plt.ylabel("Percentage (%)")
plt.title("KG-RAG Performance Across Different Dataset Sizes")
plt.ylim(0, 100)
plt.legend()
plt.grid(True, linestyle="--", alpha=0.4)

plt.tight_layout()
plt.savefig("kg_rag_dataset_sizes.png", dpi=300, bbox_inches="tight")
plt.show()
```
